```python
from bs4 import BeautifulSoup
import pandas as pd
# create dataframe with colomn ,movie name,rating,director name
file=open("30 Most Popular Movies Right Now_ What to Watch In Theaters and Streaming __ Rotten Tomatoes – Movie and TV News.html", encoding="utf-8")
responce=file.read()
soup=BeautifulSoup(responce,"html.parser")
movie=[]
rating=[]
director=[]
a=soup.find_all("div",class_="col-sm-18 col-full-xs countdown-item-content")
for i in a:
    movie.append(i.find("h2").find("a").text.strip())
    rating.append(i.find("span",class_="tMeterScore").text.strip())
    director.append(i.find("div",class_="info director").find("a").text.strip())
df=pd.DataFrame({"movie":movie,"rating":rating,"director":director})
df    
```

```python
#Question 2:
# Write a python program to scrape the following file. Extract the data regarding the given mobiles and create dataframe accordingly:-
# Link to the file :- Indeed Jobs
# Required Output

from bs4 import BeautifulSoup
import pandas as pd
# create dataframe with colomn ,movie name,rating,director name
website="indeed jobs.html"
file=open(website, encoding="utf-8")
responce=file.read()
soup=BeautifulSoup(responce,"html.parser")
name_job=[]
company=[]
salary=[]
location=[]
import numpy as np
#print(a)
a=soup.find_all("div",class_="job_seen_beacon")
for i in a:
    name_job.append(i.find("h2",class_="jobTitle css-1psdjh5 eu4oa1w0").text.strip())
    company.append(i.find("div",class_="css-1afmp4o e37uo190").text.strip())
    location.append(i.find("div",class_="css-pt3vth e37uo190").text.strip())
    try:
        salary.append(i.find("div",class_="css-18z4q2i eu4oa1w0").text.strip())
    except:
        salary.append(np.nan)    
df=pd.DataFrame({"name_job":name_job,"company":company,"location":location,"salary":salary})
df    
```
