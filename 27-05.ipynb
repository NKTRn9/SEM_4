```python
url3="https://api.openweathermap.org/data/2.5/forecast?lat=" + str(lat) + "&lon=" + str(lon) + "&appid=" + api_key+"&units=metric"
response3=requests.get(url3)
result3=response3.json()
print(json.dumps(result3,indent=3))
```

```python
date_time=[]
temp=[]
weather=[]
for i in result3['list']:
    date_time.append(i['dt_txt'])
    temp.append(i['main']['temp'])
    weather.append(i['weather'][0]['description'])
import pandas as pd
df=pd.DataFrame({'date_time':date_time,'temp':temp,'weather':weather})
df
```

```python
# Send POST request
new_post = {
"title": "Learning Requests",
"body": "This is tutorial content",
"userId": 1
}
response = requests.post(
"https://jsonplaceholder.typicode.com/posts",
json=new_post
)
print(response.status_code)
print(response.json())
```

```python
# Send PUT request
updated_post = {
"id": 1,
"title": "Updated Title",
"body": "Updated body",
"userId": 1
}
response = requests.put(
"https://jsonplaceholder.typicode.com/posts/1",
json=updated_post
)
print(response.json())
print(response.status_code)
```

```python
# Send PATCH request
patch_data = {
"title": "Patched Title"
}
response = requests.patch(
    "https://jsonplaceholder.typicode.com/posts/1",
json=patch_data
)
print(response.json())
print(response.status_code)
```

```python
# Send DELETE request
response = requests.delete(
"https://jsonplaceholder.typicode.com/posts/1"
)
print(response.status_code)
```
