```python
## Convert API response to DataFrame
import pandas as pd
response = requests.get(
"https://jsonplaceholder.typicode.com/posts"
)
posts = response.json()
df = pd.DataFrame(posts)
print(df.head())
# Loop through API data
for post in posts[:5]:
    print("ID:", post["id"])
    print("TITLE:", post["title"])
    print("BODY:", post["body"])
    print("=" * 50)
```

```python
# Get nested JSON response
response = requests.get(
"https://jsonplaceholder.typicode.com/users/1"
)
user = response.json()
print(user)
# Access nested JSON values
print(user["name"])
print(user["email"])
print(user["address"]["city"])
print(user["company"]["name"])
```

```python
# Convert JSON string into Python object
json_string = '''
{
"name": "Alice",
"age": 25
}
'''
python_data = json.loads(json_string)
print(python_data)
print(type(python_data))
```

```python
# Convert Python object into JSON string
person = {
"name": "Bob",
"age": 30
}
json_output = json.dumps(person, indent=4)
print(json_output)
```

```python
#pagination
import requests,json
url="https://jsonplaceholder.typicode.com/posts"
params = {"_page":1, "_limit":5}
r=requests.get(url, params=params)
data = r.json()
for i in data:
    print("ID:", i["id"])
    print("TITLE:", i["title"])
 
```

```python
# multiple pagination
import requests,json
url="https://jsonplaceholder.typicode.com/posts"
page=1
while page<=3:
    params = {"_page":page, "_limit":5}
    
    r=requests.get(url, params=params)
    data = r.json()
    for i in data:
        print("ID:", i["id"])
        print("TITLE:", i["title"])
    page+=1
```
